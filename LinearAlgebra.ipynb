```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t=np.linspace(0,50,100)
d_r=2.5*t
d_s=3*(t-5)
```

```python
fig,ax=plt.subplots()
plt.title("A Bank Robbery")
plt.xlabel("Time")
plt.ylabel("Distance")
ax.set_xlim(0,50)
ax.set_ylim(0,100)
ax.plot(t,d_r,c="Red")
ax.plot(t,d_s,c="Green")
plt.axvline(x=30,c="Black",linestyle="--")
plt.axhline(y=75,c="Black",linestyle="--")
plt.show()
```
